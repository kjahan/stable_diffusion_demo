```python
from diffusers import StableDiffusionPipeline
import torch

import time
```

```python
model_path = 'prompthero/midjourney-v4-diffusion'
pipe = StableDiffusionPipeline.from_pretrained(model_path , torch_dtype=torch.float16)
```

```python
!nvidia-smi
```

```python
if torch.cuda.is_available():
  pipe = pipe.to("cuda")

device = "GPU 🔥" if torch.cuda.is_available() else "CPU 🥶"

device
```

## Create images

```python
base_prompt = "Full body view of a Samurai warrior in the year 2432, 4k upscale, with London background, raining"

prompt = "mdjrny-v4 style " + base_prompt

print(prompt)
```

```python
n = 10

images = []

ts = int(time.time())

print("run time: {}".format(ts))

for i in range(n):
  result = pipe(
        prompt,
        width = 512,
        height = 512)
  image = result.images[0]

  image.save("images/sd_img_{}_{}.png".format(ts, i+1))

  images.append(image)
```

```python
fn = "images/sd_img_1700459388_{}.png"

emb_vectors = []

for i in range(n):
  img = Image.open(fn.format(str(i+1)))
  # Get a vector from img2vec, returned as a torch FloatTensor
  vec = img2vec.get_vec(img, tensor=True)
  v_ = vec.cpu().detach().numpy()
  v = np.reshape(v_, (1, 512))
  emb_vectors.append(np.squeeze(np.asarray(v)))
```

```python
len(emb_vectors)
```

```python
from numpy import dot
from numpy.linalg import norm

a = emb_vectors[3]

for i in range(n):
  b = emb_vectors[i]

  cos_sim = dot(a, b)/(norm(a)*norm(b))
  print(cos_sim)
```
