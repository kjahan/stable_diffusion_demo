## Van Gogh Diffusion

Dreambooth model trained on screenshots from the film Loving Vincent by https://huggingface.co/dallinmackay

https://huggingface.co/dallinmackay/Van-Gogh-diffusion

https://twitter.com/_akhaliq/status/1588857343181479937

```python
from diffusers import StableDiffusionPipeline
import torch
```

```python
model_id = "dallinmackay/Van-Gogh-diffusion"
pipe = StableDiffusionPipeline.from_pretrained(model_id, torch_dtype=torch.float16)
pipe = pipe.to("cuda")
```

```python
# prompt and settings used: lvngvncnt, [person], highly detailed | Steps: 25, Sampler: Euler, CFG scale: 6
prompt = "lvngvncnt, jeff bezos at sunset"
image = pipe(prompt).images[0]

image.save("sunset.png")
```

```python
image
```

```python
#If you get too many yellow faces or you dont like the strong blue bias, simply put them in the negative prompt (e.g., "Yellow face, blue").

prompt = "lvngvncnt, jeff bezos at sunset"
negative_prompt = "Yellow face, blue"
image = pipe(prompt=prompt, negative_prompt=negative_prompt).images[0]

image.save("jeff_bezos.png")
```

```python
image
```

```python
# prompt and settings used: lvngvncnt, [subject/setting], highly detailed | Steps: 25, Sampler: Euler, CFG scale: 6

prompt = "lvngvncnt, washington DC in sunrise"
image = pipe(prompt).images[0]

image.save("sunrise.png")
```
